```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
from pathlib import Path
```

```python
file_path = Path.cwd().parent.parent/"Data"/"Cleaned"/"Orders.csv"
df = pd.read_csv(file_path)
df
```

```python
df.shape
df.info()
```

```python
df["Month"] = pd.to_datetime(df["Order_Date"]).dt.month
df
```

```python
# Total Orders
total_orders = df.shape[0]
total_orders
```

```python
# Average Order value
average_order_value = df["Revenue"].mean()
average_order_value.round(2)
```

```python
df.groupby("Order_Status").agg(
    Total_Orders = ("Order_ID", "count"),
    Total_Revenue = ("Revenue", "sum"),
    Average_Order_Value = ("Revenue", "mean")
).round(2).sort_values(by="Total_Revenue", ascending=False)
```

```python
df.groupby("Customer_ID").agg(
    Total_Orders = ("Order_ID", "count"),
    Total_Revenue = ("Revenue", "sum"),
    Average_Order_Value = ("Revenue", "mean")
).round(2).sort_values(by="Total_Revenue", ascending=False).head(10)
```
