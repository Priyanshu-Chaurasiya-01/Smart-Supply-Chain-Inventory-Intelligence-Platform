```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
from pathlib import Path
```

```python
file_path = Path.cwd().parent.parent/"Data"/"Cleaned"/"Suppliers.csv"
df = pd.read_csv(file_path)
df
```

```python
df.shape
df.info()
```

```python
# Adding Reliability Category for Analysis
df["Reliability_Category"] = pd.cut(
    df["Reliability_Score"],
    bins=[0, 60, 80, 100],
    labels=["Low", "Medium", "High"]
)
```

```python
# Adding Delivery Category for Analysis
def delivery_category(x):
    if x <= 7:
        return "Fast"
    elif x <= 14:
        return "Medium"
    elif x<=21:
        return "Slow"
    else :
        return "Poor"

df["Lead_Time_Category"] = (
    df["Lead_Time_Days"].apply(delivery_category)
)
```

```python
df.info()
```

```python
# Region wise Suppliers Information
result = df.groupby("Region").agg(
    Total_Suppliers = ("Supplier_ID", "count"),
    Avg_Time_Delay = ("Lead_Time_Days", "mean"),
    Avg_Reliability_Score = ("Reliability_Score", "mean")
).round(2).reset_index().sort_values(by="Avg_Reliability_Score", ascending=False)
result
```

```python
result = df.groupby("Lead_Time_Category").agg(
    Total_Suppliers = ("Supplier_ID", "count")
)
result
```

```python
result = df.groupby("Reliability_Category").agg(
    Total_Suppliers = ("Supplier_ID", "count")
)
result
```
