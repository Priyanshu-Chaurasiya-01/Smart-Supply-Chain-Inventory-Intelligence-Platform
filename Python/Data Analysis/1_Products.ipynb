```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
from pathlib import Path
```

```python
file_path = Path.cwd().parent.parent/"Data"/"Cleaned"/"Products.csv"
df = pd.read_csv(file_path)
df
```

```python
df.shape
df.info()
```

```python
df["Profit"] = (df["Selling_Price"] - df["Unit_Cost"]).round(2)
df["Margin_Pct"] = ((df["Profit"] / df["Selling_Price"])*100).round(2)
```

```python
df.info()
```

```python
# Products with High Profit
df.sort_values(by="Profit", ascending=False).head(10)
```

```python
# Products with High Margin%
df.sort_values(by="Margin_Pct", ascending=False).head(10)
```

```python
result = df.groupby("Category").agg(
    Total_Products = ("Product_ID", "count"),
    Mean_Profit = ("Profit", "mean"),
    Mean_Margin_Pct = ("Margin_Pct", "mean")
).round(2).reset_index().sort_values(by="Mean_Profit", ascending=False)
result
```
