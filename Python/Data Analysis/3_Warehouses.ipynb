**Warehouses**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
from pathlib import Path
```

```python
file_path = Path.cwd().parent.parent/"Data"/"Cleaned"/"Warehouses.csv"
df = pd.read_csv(file_path)
df
```

```python
df.shape
df.info()
```

```python
# Adding Category according to Capacity
def capacity_category(x):
    if x < 25000:
        return "Low"
    elif x < 50000:
        return "Medium"
    else:
        return "High"
df["Capacity_Category"] = df["Capacity"].apply(capacity_category)
df
```

```python
# Adding Category according to Operating Cost
def operating_cost_category(x):
    if x < 150000:
        return "Low"
    elif x < 250000:
        return "Medium"
    else:
        return "High"
df["Operating_Cost_Category"] = df["Operating_Cost"].apply(operating_cost_category)
df
```

```python
# Warehouse having Low Capacity and High Operating Cost
df.loc[
    (df["Operating_Cost_Category"]=="High") &
    (df["Capacity_Category"]=="Low")
]
```

```python
df.to_csv("../../Data/Processed/Warehouses.csv", index=False)
```
