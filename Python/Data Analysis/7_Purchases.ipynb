```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
from pathlib import Path
```

```python
file_path = Path.cwd().parent.parent/"Data"/"Cleaned"/"Purchases.csv"
df = pd.read_csv(file_path)
df
```

```python
# Converting to date data type
cols = ["Purchase_Date", "Expected_Date", "Actual_Date"]
for col in cols:
    df[col] = pd.to_datetime(df[col], format="%Y-%m-%d", errors="coerce")
df.info()
```

```python
df
```

```python
df.shape
df.info()
```

```python
# Adding Total Bill column
df["Total_Bill"] = (df["Quantity"] * df["Unit_Cost"]).round(2)
```

```python
delivery_days = (df["Actual_Date"] - df["Purchase_Date"]).dt.days
df["Delivery_Days"] = delivery_days
```

```python
def delivery_status(row):
    if row["Actual_Date"] < row["Expected_Date"]:
        return "Early Delivery"
    elif row["Actual_Date"] == row["Expected_Date"]:
        return "On Time"
    elif row["Actual_Date"] > row["Expected_Date"]:
        return "Late Delivery"
    else:
        return np.nan
df["Delivery_Status"] = df.apply(delivery_status, axis=1)
df
```

```python
df.to_csv("../../Data/Processed/Purchases.csv", index=False)
```
