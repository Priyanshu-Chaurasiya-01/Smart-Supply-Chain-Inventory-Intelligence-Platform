```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
from pathlib import Path
```

```python
file_path = Path.cwd().parent.parent/"Excel"/"6_Customers.xlsx"
df = pd.read_excel(file_path)
df
```

```python
df.shape
df.info()
```

```python
# Capitalize, Trimming and Consistent null values
columns = ["Customer_ID", "Customer_Segment", "Region", "Acquisition_Channel"]
for col in columns:
    df[col] = df[col].str.capitalize()
    df[col] = df[col].str.strip()
nulls = ["N/a", "Na", "Null", "Unknown", ""]
df.replace(nulls, np.nan, inplace=True)
df.info()
```

```python
# Dropping duplicates except nulls
mask = df["Customer_ID"].notna()
df = pd.concat([
    df[mask].drop_duplicates(subset=["Customer_ID"], keep="first"),
    df[~mask]
]).sort_index()
df.reset_index(drop=True, inplace=True)
```

```python
# Replacing null values in Customer ID
for i in range(len(df)):
    value = df.loc[i, "Customer_ID"]
    if not pd.notna(value):
        df.loc[i, "Customer_ID"] = f"C{10000+i}"
df["Customer_ID"].isna().sum()
df["Customer_ID"].duplicated().sum()
df.info()
```
